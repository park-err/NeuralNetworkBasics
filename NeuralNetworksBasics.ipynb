## 2. Neurons

### 2.1 Neuron Formula

The sum of a bias and the dot product of a vector of inputs and a vector of weights.

$$\left( \sum_{i=1}^n x_i w_i \right) + bias$$

**Example**

```python
import numpy as np
inputs = [1.0, 2.0, 3.0, 2.5, 6.7, 4.8]
weights = [0.2, 0.8, -0.5, 0.7, 0.1, 0.4]
bias = 2.0

output = np.dot(weights, inputs) + bias
print(output)
```

### 2.2 Neuron Layering

The dot product of a vector of inputs, a matrix of weights, and a vector of biases.

$$\sum_{k=1}^n \left( \sum_{i=1}^k x_i w_ki \right) + b_k$$

**Example**

```python
import numpy as np
inputs = [1.0, 2.0, 3.0]
weights = [
    [0.2, 0.8, -0.5],
    [0.7, 0.9, -0.3]
]
biases = [2.0, 3.0]

layer_outputs = np.dot(weights, inputs) + biases
print(layer_outputs)
```
